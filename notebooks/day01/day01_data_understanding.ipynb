```python
from pathlib import Path

import pandas as pd

def find_project_root() -> Path:
    """Locate repo root (dataset/ + requirements.txt) from notebooks/dayXX/."""
    p = Path.cwd().resolve()
    for _ in range(6):
        if (p / "dataset").is_dir() and (p / "requirements.txt").exists():
            return p
        p = p.parent
    raise FileNotFoundError("Project root not found. cwd should be notebooks/dayXX/.")

PROJECT_ROOT = find_project_root()
DATA_DIR = PROJECT_ROOT / "dataset"

career_df = pd.read_csv(DATA_DIR / "AI_Career_Guidance_Dataset.csv")
skills_map_df = pd.read_csv(DATA_DIR / "Career_Skill_Mapping.csv")

career_df.shape
```
