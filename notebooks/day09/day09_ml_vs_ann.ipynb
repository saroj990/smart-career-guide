```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

def find_project_root() -> Path:
    p = Path.cwd().resolve()
    for _ in range(6):
        if (p / "dataset").is_dir() and (p / "requirements.txt").exists():
            return p
        p = p.parent
    raise FileNotFoundError("Project root not found.")

PROJECT_ROOT = find_project_root()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

sns.set_theme(style="whitegrid", context="notebook")

from src.model_comparison import build_comparison_table, pick_production_row, save_production_model
from src.preprocessing import load_cleaned_table, prepare_train_test_bundle
```

```python
bundle = prepare_train_test_bundle(load_cleaned_table(PROJECT_ROOT))
comparison = build_comparison_table(PROJECT_ROOT)
comparison
```

```python
winner = pick_production_row(comparison)
print("Selected:", winner["model"], "F1 macro", winner["test_f1_macro"])
paths = save_production_model(
    comparison,
    PROJECT_ROOT,
    X_train=bundle["X_train"],
    y_train=bundle["y_train"],
    career_classes=bundle["career_classes"],
)
paths
```
